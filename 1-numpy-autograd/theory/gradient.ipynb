## Chain Rule

### Single variable

#### Definition

if $y = f(g(x))$, then:
$$
\frac{dy}{dx} = f^{\prime}(g(x)) \cdot g^{\prime}(x)
$$

with intermediate variable $u = g(x)$:
$$
\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}
$$

#### Proof

Let $u = g(x)$, so $y = f(u)$. By definition of the derivative:

$$
\frac{dy}{dx} = \lim_{h \to 0} \frac{f(g(x + h)) - f(g(x))}{h}
$$

Let $k = g(x + h) - g(x)$, i.e. how much $u$ changes when $x$ changes by $h$. Multiply and divide by $k$:

$$
\frac{dy}{dx} = \lim_{h \to 0} \frac{f(u + k) - f(u)}{k} \cdot \frac{k}{h}
$$

As $h \to 0$, we also have $k \to 0$ (since $g$ is continuous), so:

$$
\frac{dy}{dx} = \underbrace{\lim_{k \to 0} \frac{f(u + k) - f(u)}{k}}_{f'(u) \,=\, f'(g(x))} \cdot \underbrace{\lim_{h \to 0} \frac{k}{h}}_{g'(x)}
$$

$$
\boxed{\frac{dy}{dx} = f'(g(x)) \cdot g'(x)}
$$

### Multiple variables

#### Definition

if $z = f(u, v)$ where $u = g(x, y)$ and $v = h(x, y)$, then:
$$
\frac{\partial z}{\partial x} = \frac{\partial z}{\partial u} \cdot \frac{\partial u}{\partial x} + \frac{\partial z}{\partial v} \cdot \frac{\partial v}{\partial x}
$$

#### Proof

Let $\Delta u = g(x+k) - g(x)$ and $\Delta v = h(x+k) - h(x)$, then by definition:

$$
\frac{\partial z}{\partial x} = \lim_{k \to 0} \frac{f(u + \Delta u,\ v + \Delta v) - f(u, v)}{k}
$$

Add and subtract $f(u + \Delta u,\ v)$ in the numerator:

$$
= \lim_{k \to 0} \frac{f(u + \Delta u,\ v + \Delta v) - f(u + \Delta u,\ v) + f(u + \Delta u,\ v) - f(u,\ v)}{k}
$$

Split into two fractions:

$$
= \lim_{k \to 0} \frac{f(u + \Delta u,\ v + \Delta v) - f(u + \Delta u,\ v)}{k} + \lim_{k \to 0}\frac{f(u + \Delta u,\ v) - f(u,\ v)}{k}
$$

Multiply and divide each term:

$$
= \lim_{k \to 0} \frac{f(u + \Delta u,\ v + \Delta v) - f(u + \Delta u,\ v)}{\Delta v} \cdot \frac{\Delta v}{k} + \lim_{k \to 0} \frac{f(u + \Delta u,\ v) - f(u,\ v)}{\Delta u} \cdot \frac{\Delta u}{k}
$$

As $k \to 0$, $\Delta u \to 0$ and $\Delta v \to 0$:

$$
\boxed{\frac{\partial z}{\partial x} = \frac{\partial z}{\partial u} \cdot \frac{\partial u}{\partial x} + \frac{\partial z}{\partial v} \cdot \frac{\partial v}{\partial x}}
$$

## Gradient

#### Definition

for a scalar function $f : \mathbb{R} ^n \to \mathbb{R}$, the gradient is:

$$
\Delta f(x) = 
\begin{bmatrix} 
\frac{\partial f}{\partial x_1} \\
\frac{\partial f}{\partial x_2} \\
\vdots \\
\frac{\partial f}{\partial x_n} \\
\end{bmatrix}
$$

- it takes a point $\mathbf{x} \in \mathbb{R}^n$ and returns a vector in $\mathbb{R}^n$
- The gradient $\Delta f$ at a point is a 2D vector sitting in the $(x, y)$ plane. It points in the direction where function $f$ increases by the largest amount.



---

## Jacobian



For a function $\mathbf{f}: \mathbb{R}^n \to \mathbb{R}^m$, the Jacobian is:

$$J = \begin{bmatrix} \frac{\partial f_1}{\partial x_1} & \cdots & \frac{\partial f_1}{\partial x_n} \\ \vdots & \ddots & \vdots \\ \frac{\partial f_m}{\partial x_1} & \cdots & \frac{\partial f_m}{\partial x_n} \end{bmatrix} \in \mathbb{R}^{m \times n}$$

Entry $J_{ij}$ answers: how much does output $i$ change when input $j$ is nudged?


## Chain rule (vector form)

If $\mathbf{y} = f(\mathbf{u})$ and $\mathbf{u} = g(\mathbf{x})$, then:

$$\frac{\partial \mathbf{y}}{\partial \mathbf{x}} = J_f \cdot J_g$$

The chain rule is just matrix multiplication of Jacobians.

## Relation to ML

let's create a model for `MNIST` with 2 linear layers: 

$$
l_1 : \mathbb{R}^{28 \times 28} \to \mathbb{R}^{64}
$$
$$
l_2 : \mathbb{R}^{64} \to \mathbb{R}^{10}
$$
$$
\mathcal{L} : \mathbb{R}^{10} \to \mathbb{R}
$$

where 
$$
l_1(x \mid W_1, B_1) = x W_1 + B_1
$$
$$
l_2(x \mid W_2, B_2) = x W_2 + B_2
$$
$$
\mathcal{L}(x, y) = \frac{1}{n} \sum_{i=1}^{n}{(y_i - x_i) ^ 2}
$$

so output is `output = loss_fn(l_2(l_1(input)))`

and propagation function is 
$$
f(x, W_1, W_2, B_1, B_2, \mathcal{L} \mid y) = \mathcal{L}(W_2 (W_1 x + B_1) + B_2 \mid y)
$$
where $f : \mathbb{R} ^ {28 \times 28} \to \mathbb{R}$


then, we need to calculate gradient of the loss function with respect to each parameter:

$$
\Delta f= 
\begin{bmatrix} 
\frac{\partial \mathcal{L}}{\partial W_1} \\
\frac{\partial \mathcal{L}}{\partial W_2} \\
\frac{\partial \mathcal{L}}{\partial B_1} \\
\frac{\partial \mathcal{L}}{\partial B_2} \\
\end{bmatrix} 
$$


$$
\frac{\partial \mathcal{L}}{\partial B_2} = J_{\mathcal{L}} \cdot J_{B_2}
$$

$$
\frac{\partial \mathcal{L}}{\partial W_2} = \frac{\partial \mathcal{L}}{\partial B_2} \cdot \frac{\partial B_2}{\partial W_2} = J_{\mathcal{L}} \cdot J_{B_2} \cdot J_{B_2} \cdot J_{W_2} 
$$

$$
\frac{\partial \mathcal{L}}{\partial B_1} = \frac{\partial \mathcal{L}}{\partial B_2} \cdot \frac{\partial B_2}{\partial W_2} \cdot \frac{\partial W_2}{\partial B_1} = J_{\mathcal{L}} \cdot J_{B_2} \cdot J_{B_2} \cdot J_{W_2} \cdot J_{W_2} \cdot J_{B_1} 
$$

$$
\frac{\partial \mathcal{L}}{\partial W_1} = \frac{\partial \mathcal{L}}{\partial B_2} \cdot \frac{\partial B_2}{\partial W_2} \cdot \frac{\partial W_2}{\partial B_1} \cdot \frac{\partial B_1}{\partial W_1} = J_{\mathcal{L}} \cdot J_{B_2} \cdot J_{B_2} \cdot J_{W_2} \cdot J_{W_2} \cdot J_{B_1} \cdot J_{B_1} \cdot J_{W_1}
$$

### Manual Calculation

let \
$x$ - Input (the MNIST dataset, $\mathbb{R}^{784}$) \
$h$ - Hidden layer output (after first linear layer, $\mathbb{R}^{64}$) \
$o$ - Output (after second linear layer, $\mathbb{R}^{10}$) \
$\mathcal{L}$ - Loss (scalar)
weights: \

$A, \mathbf{a}$ - First layer weights and bias \
$B, \mathbf{b}$ - Second layer weights and bias

Then **forward pass** is :
$$
h = x A + \mathbb{a}
$$
$$
o = h B + \mathbb{b}
$$
$$
\mathcal{L} = MSE(o, y)
$$

##### Gradient of $\mathcal{L}$ w.r.t. $o$
$$
\delta = \frac{\partial \mathcal{L}}{\partial o} = \frac{1}{n}(o - y)
$$

##### Ggradient of $\mathcal{L}$ w.r.t. B
as $o = h B + \mathbb{b}$, element by element:
$$
o_j = \sum_{j}{h_i \cdot B_{ij}}
$$
nudging $B_{ij}$ only affects $o_j$, so:
$$
\frac{\partial \mathcal{L}}{\partial B_{ij}} = \frac{\partial \mathcal{L}}{\partial o_j} \cdot \frac{\partial o_j}{\partial B_{ij}} = \delta_j \cdot h_i
$$
in matrix form:
$$
\boxed{
    \frac{\partial \mathcal{L}}{\partial B_{ij}} = h ^T \delta \in \mathbb{R}^{64 \times 10}
}
$$

##### Gradient w.r.t. b
$$
\boxed{
    \frac{\partial \mathcal{L}}{\partial \mathbb{b}} = \delta \in \mathbb{R}^{10}
}
$$
