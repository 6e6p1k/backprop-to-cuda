```python
import bobrito
from bobrito import Tensor
from bobrito import bobr
import torch
import time
from contextlib import contextmanager

@contextmanager
def timing(label=''):
    start = time.perf_counter()
    yield
    end = time.perf_counter()
    print(f"{label+': ' if label else ''}{end-start:.8f}s")
```

```python
a = Tensor([1, 2, 3], [3])
a.device = "cuda"
print(a.device)
```

## matmul testing section

```python
a = bobr.randn(128, 52, 64, 32, device="cuda")
b = bobr.randn(128, 52, 32, 128, device="cuda")

a_t = a.torch()
b_t = b.torch()
print(a.device, b.device)
```

```python
%%timeit
c = a.matmul_fast(b)
```

```python
%%timeit
c = a_t @ b_t
torch.cuda.synchronize()
```

## add testing section

```python
a = bobr.randn(128, 32, 128, device="cuda")
b = bobr.randn(128, 32, 128, device="cuda")

a_t = a.torch()
b_t = b.torch()
print(a.device, b.device)
```

```python
with timing("add on cuda"):
    for i in range(100):
        c = a + b
```

```python
with timing("tortik add"):
    for i in range(100):
        c = a_t + b_t
```

```python
with timing("mul on cuda"):
    for i in range(100):
        c = a + b
```

```python
with timing("tortik mul"):
    for i in range(100):
        c = a_t + b_t
```

## SUM TEST

```python
a = bobr.randn(512, 512, 512, device="cuda")
a_t = a.torch()
```

```python
with timing("SUM on cuda"):
    for i in range(100):
        c = a.sum(0)
        c = a.sum(1)
        c = a.sum(2)
```

```python
with timing("tortik sum"):
    for i in range(100):
        c = a_t.sum(0)
        c = a_t.sum(1)
        c = a_t.sum(2)
```
